```python
import torch
from torch import nn
import torchvision
from torchvision import transforms
from torchvision import datasets
from torchvision.transforms import ToTensor
import matplotlib.pyplot as plt
```

```python
class FashionMNISTModel(nn.Module):
    def __init__(self,
            input_shape: int,
            hidden_shape: int,
            output_shape: int):
        super().__init__()

        self.conv_block_1=nn.Sequential(
            nn.Conv2d(in_channels=input_shape,out_channels=hidden_shape,kernel_size=3,stride=1,padding=1),
            nn.ReLU(),
            nn.Conv2d(in_channels=hidden_shape,out_channels=hidden_shape,kernel_size=3,stride=1,padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2)
        )

        self.conv_block_2=nn.Sequential(
            nn.Conv2d(in_channels=hidden_shape,out_channels=hidden_shape,kernel_size=3,stride=1,padding=1),
            nn.ReLU(),
            nn.Conv2d(in_channels=hidden_shape,out_channels=hidden_shape,kernel_size=3,stride=1,padding=1),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2)
        )

        self.classifier=nn.Sequential(
            nn.Flatten(),
            nn.Linear(in_features=hidden_shape*0,out_features=output_shape)
        )

    def forward(self,x):
        x=self.conv_block_1(x)
        print(x.shape)
        x=self.conv_block_2(x)
        print(x.shape)
        x=self.classifier(x)
        return x
```
