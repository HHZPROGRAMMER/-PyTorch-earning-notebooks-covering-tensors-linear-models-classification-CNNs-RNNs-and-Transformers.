```python
import torch 
import pandas as pd
import numpy as np
import torch.nn as nn
```

```python
torch.manual_seed(42)

input_size=3
hidden_size=4
squence_length=5
batch_size=2

x=torch.randn(
    batch_size,
    squence_length,
    input_size
)

print(x.shape)

w_xh=nn.Parameter(
    torch.randn(input_size,hidden_size)*0.1
)

w_hh=nn.Parameter(
    torch.randn(hidden_size,hidden_size)*0.1
)

b_h=nn.Parameter(
    torch.zeros(hidden_size)
)

print("w_xh:",w_xh.shape)
print("w_hh:",w_hh.shape)
print("Bias:",b_h.shape)

h=torch.zeros(batch_size,hidden_size)

print(h.shape)
```

```python
hidden_states = []

for t in range(squence_length):

    x_t = x[:, t, :]

    input_part = x_t @ w_xh

    hidden_part = h @ w_hh

    h = torch.tanh(
        input_part + hidden_part + b_h
    )

    hidden_states.append(h)

    print(f"Step {t + 1}")
    print("Input shape:", x_t.shape)
    print("Hidden shape:", h.shape)
    print()
```

```python
outputs = torch.stack(hidden_states, dim=1)

print(outputs.shape)

final_hidden = outputs[:, -1, :]

print(final_hidden.shape)
```

```python
class RNN_STT(nn.Module):

    def __init__(
            self,
            input_size=80,
            hidden_size=256,
            vocab_size=45
    ):
        super().__init__()

        self.rnn=nn.RNN(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True,
            bidirectional=True
        )

        self.fc=nn.Linear(
            hidden_size*2,
            vocab_size
        )

    def forward(self,x):

        output, hidden=self.rnn(x)
        logits=self.fc(output)
        return logits
```

```python
model=RNN_STT()

x=torch.randn(2,200,80)

logits=model(x)
print(logits.shape)
```
