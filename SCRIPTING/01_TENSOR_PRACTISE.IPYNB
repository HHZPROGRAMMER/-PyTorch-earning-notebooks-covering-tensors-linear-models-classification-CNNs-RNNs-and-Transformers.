```python
import torch
from torch import nn

torch.manual_seed(42)

class Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.layers=nn.Sequential(
            nn.Linear(8,16),
            nn.ReLU(),
            nn.Linear(16,32),
            nn.ReLU(),
            nn.Linear(32,2)
        )

    def forward(self,x):
        return self.layers(x)
```

```python
X=torch.randn(24,8)

model=Model()

output=model(X)

output.shape
```

```python
import torch

x=torch.tensor(2.0)
target=torch.tensor(10.0)

w=torch.tensor(3.0,requires_grad=True)
b=torch.tensor(1.0,requires_grad=True)

prediction=x*w+b

loss=(prediction-target)**2

print(f"Loss: {loss.item()}")
print(f"Prediction: {prediction.item()}")

loss.backward()

print("Updated w: ",w.grad.item())
print("Updated b:",b.grad.item())
```

```python
learning_rate=0.05

with torch.no_grad():
    w-=learning_rate*w.grad
    b-=learning_rate*b.grad

print("New weight: ",w.item())
print("New bias: ",b.item())

prediction=x*w+b

loss=(prediction-target)**2

print(f"Loss: {loss.item()}")
print(f"Prediction: {prediction.item()}")

loss.backward()

print("Updated w: ",w.grad.item())
print("Updated b:",b.grad.item())
```

```python
X=torch.arange(1,6,dtype=torch.float32).unsqueeze(1)

y=X*2+1

model=nn.Linear(1,1)

loss_fn=nn.MSELoss()

optimizer=torch.optim.SGD(model.parameters(),lr=0.01)

for epoc in range(500):
    model.train()

    optimizer.zero_grad()

    prediction=model(X)

    loss=loss_fn(prediction,y)

    loss.backward()

    optimizer.step()

    if (epoc+1)%100==0:
        print(f"Epoc: {epoc+1}  |  Loss: {loss.item():.6f}")

X.shape
```

```python

x=torch.tensor([[1.],[2.],[3.],[4.],[5]])
y=torch.tensor([[5.],[8.],[11.],[14.],[17.]])

w=torch.tensor(1.,requires_grad=True)
b=torch.tensor(1.,requires_grad=True)

learning_rate=0.01

epocs=20

for epoc in range(epocs):

    prediction=w*x+b

    loss=((prediction-y)**2).mean()

    loss.backward()

    with torch.no_grad():
        w-=learning_rate*w.grad
        b-=learning_rate*b.grad

    w.grad.zero_()
    b.grad.zero_()

    if epoc%4==0:
        print(f"Epoc: {epoc}   | Loss: {loss}")

```

```python
X=torch.randn(32,8)

class Model(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer_1=nn.Linear(8,64)
        self.relu=nn.ReLU()
        self.layer_2=nn.Linear(64,32)
        self.layer_3=nn.Linear(32,16)
        self.layer_4=nn.Linear(16,4)

    def forward(self,X):
        return self.layer_4(self.relu(self.layer_3(self.relu(self.layer_2(self.relu(self.layer_1(X)))))))

model=Model()
print(model.named_parameters())
print(model)
```

```python
class ModelSequential(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer=nn.Sequential(
            nn.Linear(8,64),
            nn.ReLU(),
            nn.Linear(64,32),
            nn.ReLU(),
            nn.Linear(32,16),
            nn.ReLU(),
            nn.Linear(16,4)
        )

    def forward(self,X):
        return self.layer(X)
```

```python
model_sequential=ModelSequential()
print(model_sequential(X).shape)
```

```python
model_sequential.eval()
with torch.inference_mode():
    prediction=model_sequential(X)
print(f"ModelSequential: {prediction.shape}")
```

```python
import torch
from torch.utils.data import TensorDataset,random_split

X=torch.randn(1000,5)

y=(2*X[:,0]+3*X[:,1]-X[:,2]+4*X[:,3]+0.5*X[:,4]).unsqueeze(1)

dataset=TensorDataset(X,y)
```

```python
train_size=int(0.8*len(dataset))

val_size=len(dataset)-train_size

train_dataset, val_dataset=random_split(dataset,[train_size,val_size],generator=torch.Generator().manual_seed(42))

print("Train:",len(train_dataset))
print("Validation:",len(val_dataset))
```

```python
from torch.utils.data import DataLoader

train_daloader=DataLoader(train_dataset,batch_size=32,shuffle=True)
val_daloader=DataLoader(val_dataset,batch_size=32,shuffle=True)

print("Train batches:",len(train_daloader))
print("Validation batches:",len(val_daloader))
```
