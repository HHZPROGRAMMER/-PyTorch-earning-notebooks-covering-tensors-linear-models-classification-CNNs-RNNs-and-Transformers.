```python
import torch
import torch.nn as nn
import torch.nn.functional as F

X=torch.tensor([
    [1., 0., 1., 0.],
    [0., 1., 0., 1.],
    [1., 1., 0., 0.]
])

print(X.shape)
```

```python
d_model=4

w_q=nn.Linear(4,4,bias=False)
w_k=nn.Linear(4,4,bias=False)
w_v=nn.Linear(4,4,bias=False)

Q=w_q(X)
K=w_k(X)
V=w_v(X)

print("Q:",Q.shape)
print("K:",K.shape)
print("V:",V.shape)
```

```python
import math
scores=Q@K.T/math.sqrt(d_model)
print(scores.shape)
print(scores)
```

```python
attention_weight=F.softmax(scores)
print(attention_weight)
```

```python
output=attention_weight@V
print(output.shape)
```
